# 69. The hardened phase on the 3B: one recipe on every model

**Why this notebook.** Notebook 68b gave the 7B its final model, v9: v6 (after the processing phase) continued on the hardened phase. Against v5, v9 lowered every attack measure (adaptive 0.076 to 0.005, combined 0.286 to 0.248, second source 0.087 to 0.047), raised SecFid security and fidelity, and kept the user's instructions, MMLU, copying and normal summaries. The 3B currently ends with a different last phase (notebook 67's rebalancing, v7). This notebook gives the 3B the 7B's last phase, unchanged, so that the paper presents one recipe: the processing phase, then the hardened phase.

**What is trained.** Exactly notebook 68b's phase, from the 3B's v6 (`data/distill6/lora_3b`):
1. *Mining.* The adaptive attacker of notebooks 43 to 45 (the Qwen 2.5 7B rewriter) is run against the 3B's v6 on the same 600 items of the project's training split as notebook 68: the original wording, then two rounds of three rewrites for every item v6 still resists. No BIPIA text, goal or category, no SecFid text and no evaluation item is used, and this is asserted.
2. *The set.* 1,500 attacked examples (every attempt that made v6 obey first, then the latest failed rewrites), 500 twins of the mined wordings, the 250 literal and 250 processing examples of the rebalancing set, and 500 clean-summary anchors. Every teacher prompt uses the item's own clean passage, recovered from its record.
3. *Training.* The training step of notebook 68b (`src/hardened_phase.py`): every sampled reply is scored, however short. The new model is the 3B v9.

**Evaluation.** As notebook 68b: the fixed attacker (13 wordings, the paper rule), the adaptive attacker (13 attempts), the combined measure, the second source, user-issued instructions, held-out Alpaca, MMLU, the copy battery, SecFid in notebook 62's harness, and the summary check on 60 different validation passages. The undefended 3B, v4 (the paper's 3B), v6 and v7 (notebook 67's model) are read from their caches.

**Marks, fixed in Stage A before any mining, training or reading.** Against v4, the 3B before the processing phase, as for every model; the same marks as notebook 68b.

| mark | what must hold |
